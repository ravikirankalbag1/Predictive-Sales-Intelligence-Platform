## Loading The Test Dataset

```python
import pandas as pd
import joblib
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
valid_df = pd.read_csv("../data/ml_ready/valid.csv")
test_df = pd.read_csv("../data/ml_ready/test.csv")
```

```python
print(valid_df.shape)
print(test_df.shape)
```

## Loading the Scalar & Models

```python
loaded_scaler = joblib.load("../models/scaler.joblib")
rf_loaded_models = joblib.load("../models/random_forest_model.joblib")
light_gbm_loaded_models = joblib.load("../models/light_gbm_model.joblib")
```

## Test Data Verification

```python
X_test = test_df.drop("next_monthly_revenue", axis=1)
y_test = test_df["next_monthly_revenue"]
```

```python
X_test.shape
```

```python
X_test_scaled = loaded_scaler.transform(X_test)
```

```python
rf_y_test_pred = rf_loaded_models.predict(X_test_scaled)
lightgbm_y_test_pred = light_gbm_loaded_models.predict(X_test_scaled)
```

```python
print("MAE", mean_absolute_error(y_test, rf_y_test_pred))
print("RMSE", np.sqrt(mean_squared_error(y_test, rf_y_test_pred)))
print("R2 Score", r2_score(y_test, rf_y_test_pred))


print("MAE", mean_absolute_error(y_test, lightgbm_y_test_pred))
print("RMSE", np.sqrt(mean_squared_error(y_test, lightgbm_y_test_pred)))
print("R2 Score", r2_score(y_test, lightgbm_y_test_pred))
```
