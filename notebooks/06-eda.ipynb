```python
from sqlalchemy import create_engine
import pandas as pd
```

```python
engine = create_engine("postgresql+psycopg2://postgres:1234@localhost:5432/abc_retail_db")
```

```python
conn = engine.connect()
print("Connection Successful!")
```

```python
customer_df = pd.read_sql_query("select * from customers", conn)
products_df = pd.read_sql_query("select * from products", conn)
stores_df = pd.read_sql_query("select * from stores", conn)
sales_transaction_df = pd.read_sql_query("select * from sales_transaction", conn)
dates_df = pd.read_sql_query("select * from dates", conn)
```
